```python
# =====================================================================
# PORTFOLIO 3: COMPUTER VISION - VAE ON CELEBA
# Complete Implementation: Tasks 1, 2, and 3
# =====================================================================
# 
# BEFORE RUNNING:
# 1. Add CelebA dataset: Click "+ Add Input" → Search "celeba-dataset" → Add
# 2. Enable GPU: Settings → Accelerator → GPU T4 x2
#
# TEAM MEMBERS (UPDATE THESE):
# - Member 1: [Name] - Task 1 (Baseline VAE)
# - Member 2: [Name] - Task 2 (Latent Space Exploration)  
# - Member 3: [Name] - Task 3 (β-VAE Variant)
# =====================================================================

# %% [markdown]
# # Portfolio 3: Computer Vision - VAE on CelebA
# ## Complete Implementation: Tasks 1, 2, and 3
# 
# ### Team Members:
# - **Member 1**: [Name] - Task 1 (Baseline VAE Implementation)
# - **Member 2**: [Name] - Task 2 (Latent Space Exploration)
# - **Member 3**: [Name] - Task 3 (β-VAE Variant)

# %% [markdown]
# ---
# # SETUP AND CONFIGURATION
# ---

# %%
# Check GPU availability
import torch
print("="*50)
print("SYSTEM CHECK")
print("="*50)
print(f"PyTorch version: {torch.__version__}")
print(f"CUDA available: {torch.cuda.is_available()}")
if torch.cuda.is_available():
    print(f"GPU: {torch.cuda.get_device_name(0)}")
    print(f"GPU Memory: {torch.cuda.get_device_properties(0).total_memory / 1e9:.1f} GB")
else:
    print("⚠️ WARNING: No GPU detected! Training will be very slow.")
    print("Go to Settings → Accelerator → Select GPU")
print("="*50)

# %%
# All imports
import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms
from torchvision.utils import make_grid, save_image
import matplotlib.pyplot as plt
import numpy as np
from PIL import Image
import os
from tqdm.notebook import tqdm  # Use notebook version for Kaggle
import glob
import json
from IPython.display import display, clear_output

# Set random seeds for reproducibility
torch.manual_seed(42)
np.random.seed(42)
if torch.cuda.is_available():
    torch.cuda.manual_seed(42)

# Device
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"\nUsing device: {device}")

# %%
# FIND THE CELEBA DATASET PATH
print("="*50)
print("FINDING CELEBA DATASET")
print("="*50)

# List what's in the input directory
print("\nContents of /kaggle/input/:")
for item in os.listdir('/kaggle/input/'):
    print(f"  📁 {item}")

# Try to find the images
possible_paths = [
    '/kaggle/input/celeba-dataset/img_align_celeba/img_align_celeba',
    '/kaggle/input/celeba-dataset/img_align_celeba',
    '/kaggle/input/celeba/img_align_celeba',
]

DATA_PATH = None
for path in possible_paths:
    if os.path.exists(path):
        jpg_files = glob.glob(os.path.join(path, '*.jpg'))
        if len(jpg_files) > 0:
            DATA_PATH = path
            print(f"\n✅ Found CelebA images at: {DATA_PATH}")
            print(f"   Total images: {len(jpg_files):,}")
            break

if DATA_PATH is None:
    print("\n❌ ERROR: Could not find CelebA images!")
    print("Make sure you added the dataset:")
    print("  1. Click '+ Add Input' on the right")
    print("  2. Search 'celeba-dataset'")
    print("  3. Add the dataset by Jessica Li")
    
    # Try to find any jpg files
    print("\nSearching for jpg files...")
    result = os.popen('find /kaggle/input -name "*.jpg" 2>/dev/null | head -5').read()
    print(result)

# %%
# Create output directories
os.makedirs('/kaggle/working/results/task1', exist_ok=True)
os.makedirs('/kaggle/working/results/task2', exist_ok=True)
os.makedirs('/kaggle/working/results/task3', exist_ok=True)
os.makedirs('/kaggle/working/models', exist_ok=True)
print("✅ Output directories created!")

# %% [markdown]
# ---
# # HYPERPARAMETERS
# ---

# %%
# ===========================================
# HYPERPARAMETERS - You can adjust these
# ===========================================

IMG_SIZE = 64           # Image size (64x64) - smaller = faster training
LATENT_DIM = 128        # Latent space dimension
BATCH_SIZE = 128        # Batch size (reduce if out of memory)
NUM_WORKERS = 2         # Data loading workers

# Task 1 settings
TASK1_EPOCHS = 15       # Training epochs for baseline VAE
LEARNING_RATE = 1e-4    # Learning rate

# Task 3 settings (β-VAE)
TASK3_EPOCHS = 10       # Epochs for each β-VAE variant
BETA_VALUES = [4, 10]   # β values to test (β=1 is baseline)

# Dataset limit (set to None for full dataset, or a number for quick testing)
DATASET_LIMIT = None    # Use None for full dataset, 10000 for quick test

print("="*50)
print("HYPERPARAMETERS")
print("="*50)
print(f"Image size: {IMG_SIZE}x{IMG_SIZE}")
print(f"Latent dimension: {LATENT_DIM}")
print(f"Batch size: {BATCH_SIZE}")
print(f"Task 1 epochs: {TASK1_EPOCHS}")
print(f"Task 3 epochs: {TASK3_EPOCHS}")
print(f"β values for Task 3: {BETA_VALUES}")
print(f"Dataset limit: {DATASET_LIMIT if DATASET_LIMIT else 'Full dataset'}")
print("="*50)

# %% [markdown]
# ---
# # DATASET CLASS
# ---

# %%
class CelebADataset(Dataset):
    """
    Custom Dataset for CelebA images.
    
    The CelebA dataset contains 202,599 celebrity face images.
    Images are aligned and cropped to focus on faces.
    """
    
    def __init__(self, root_dir, transform=None, limit=None):
        """
        Args:
            root_dir: Path to directory containing jpg images
            transform: Torchvision transforms to apply
            limit: Optional limit on number of images (for testing)
        """
        self.root_dir = root_dir
        self.transform = transform
        
        # Get all jpg files
        self.image_files = sorted(glob.glob(os.path.join(root_dir, '*.jpg')))
        
        # Apply limit if specified
        if limit is not None:
            self.image_files = self.image_files[:limit]
        
        print(f"CelebADataset initialized:")
        print(f"  - Path: {root_dir}")
        print(f"  - Images: {len(self.image_files):,}")
    
    def __len__(self):
        return len(self.image_files)
    
    def __getitem__(self, idx):
        # Load image
        img_path = self.image_files[idx]
        image = Image.open(img_path).convert('RGB')
        
        # Apply transforms
        if self.transform:
            image = self.transform(image)
        
        return image, idx  # Return index as label

# %%
# Define image transforms
# CelebA original images are 178x218, we resize to 64x64
transform = transforms.Compose([
    transforms.Resize((IMG_SIZE, IMG_SIZE)),    # Resize to square
    transforms.ToTensor(),                       # Convert to tensor [0, 1]
    transforms.Normalize([0.5, 0.5, 0.5],       # Normalize to [-1, 1]
                        [0.5, 0.5, 0.5])
])

print("Transform pipeline:")
print("  1. Resize to 64x64")
print("  2. Convert to tensor [0, 1]")
print("  3. Normalize to [-1, 1]")

# %%
# Load dataset and create data loaders
print("\n" + "="*50)
print("LOADING DATASET")
print("="*50)

full_dataset = CelebADataset(
    root_dir=DATA_PATH,
    transform=transform,
    limit=DATASET_LIMIT
)

# Split: 90% train, 10% test
train_size = int(0.9 * len(full_dataset))
test_size = len(full_dataset) - train_size

train_dataset, test_dataset = torch.utils.data.random_split(
    full_dataset,
    [train_size, test_size],
    generator=torch.Generator().manual_seed(42)  # Reproducible split
)

# Create data loaders
train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=NUM_WORKERS,
    pin_memory=True
)

test_loader = DataLoader(
    test_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=NUM_WORKERS,
    pin_memory=True
)

print(f"\nDataset split:")
print(f"  Training: {len(train_dataset):,} images ({len(train_loader)} batches)")
print(f"  Test: {len(test_dataset):,} images ({len(test_loader)} batches)")

# %%
# Visualize sample batch
print("\nSample batch visualization:")
sample_batch, _ = next(iter(train_loader))
print(f"Batch shape: {sample_batch.shape}")  # Should be [128, 3, 64, 64]

# Display grid
def denormalize(tensor):
    """Convert from [-1,1] to [0,1]"""
    return (tensor + 1) / 2

grid = make_grid(denormalize(sample_batch[:16]), nrow=4)
plt.figure(figsize=(8, 8))
plt.imshow(grid.permute(1, 2, 0).cpu().numpy())
plt.axis('off')
plt.title('Sample Training Images (16 faces)')
plt.savefig('/kaggle/working/results/task1/sample_batch.png', dpi=150, bbox_inches='tight')
plt.show()

# %% [markdown]
# ---
# # MODEL ARCHITECTURE
# ---

# %%
class Encoder(nn.Module):
    """
    Encoder network: Image → Latent distribution parameters (μ, σ)
    
    Architecture:
    - 4 convolutional layers with stride 2 (downsample by 2x each)
    - BatchNorm + LeakyReLU activation
    - Final FC layers for μ and log(σ²)
    
    Input: (batch, 3, 64, 64)
    Output: μ (batch, latent_dim), logvar (batch, latent_dim)
    """
    
    def __init__(self, latent_dim=128):
        super(Encoder, self).__init__()
        
        # Conv layers: 64→32→16→8→4
        self.conv1 = nn.Conv2d(3, 32, kernel_size=4, stride=2, padding=1)    # 64→32
        self.conv2 = nn.Conv2d(32, 64, kernel_size=4, stride=2, padding=1)   # 32→16
        self.conv3 = nn.Conv2d(64, 128, kernel_size=4, stride=2, padding=1)  # 16→8
        self.conv4 = nn.Conv2d(128, 256, kernel_size=4, stride=2, padding=1) # 8→4
        
        # Batch normalization
        self.bn1 = nn.BatchNorm2d(32)
        self.bn2 = nn.BatchNorm2d(64)
        self.bn3 = nn.BatchNorm2d(128)
        self.bn4 = nn.BatchNorm2d(256)
        
        # Flatten size: 256 * 4 * 4 = 4096
        self.flatten_size = 256 * 4 * 4
        
        # FC layers for μ and log(σ²)
        self.fc_mu = nn.Linear(self.flatten_size, latent_dim)
        self.fc_logvar = nn.Linear(self.flatten_size, latent_dim)
        
    def forward(self, x):
        # Conv layers with LeakyReLU
        x = F.leaky_relu(self.bn1(self.conv1(x)), 0.2)
        x = F.leaky_relu(self.bn2(self.conv2(x)), 0.2)
        x = F.leaky_relu(self.bn3(self.conv3(x)), 0.2)
        x = F.leaky_relu(self.bn4(self.conv4(x)), 0.2)
        
        # Flatten
        x = x.view(x.size(0), -1)
        
        # Get μ and log(σ²)
        mu = self.fc_mu(x)
        logvar = self.fc_logvar(x)
        
        return mu, logvar


class Decoder(nn.Module):
    """
    Decoder network: Latent vector → Reconstructed image
    
    Architecture:
    - FC layer to initial feature map
    - 4 transposed conv layers (upsample by 2x each)
    - BatchNorm + ReLU activation
    - Tanh output (to match [-1,1] normalization)
    
    Input: z (batch, latent_dim)
    Output: image (batch, 3, 64, 64)
    """
    
    def __init__(self, latent_dim=128):
        super(Decoder, self).__init__()
        
        self.latent_dim = latent_dim
        self.initial_size = 4
        
        # FC to initial feature map
        self.fc = nn.Linear(latent_dim, 256 * 4 * 4)
        
        # Transposed conv layers: 4→8→16→32→64
        self.deconv1 = nn.ConvTranspose2d(256, 128, kernel_size=4, stride=2, padding=1)  # 4→8
        self.deconv2 = nn.ConvTranspose2d(128, 64, kernel_size=4, stride=2, padding=1)   # 8→16
        self.deconv3 = nn.ConvTranspose2d(64, 32, kernel_size=4, stride=2, padding=1)    # 16→32
        self.deconv4 = nn.ConvTranspose2d(32, 3, kernel_size=4, stride=2, padding=1)     # 32→64
        
        # Batch normalization
        self.bn1 = nn.BatchNorm2d(128)
        self.bn2 = nn.BatchNorm2d(64)
        self.bn3 = nn.BatchNorm2d(32)
        
    def forward(self, z):
        # FC and reshape
        x = self.fc(z)
        x = x.view(x.size(0), 256, 4, 4)
        
        # Transposed conv with ReLU
        x = F.relu(self.bn1(self.deconv1(x)))
        x = F.relu(self.bn2(self.deconv2(x)))
        x = F.relu(self.bn3(self.deconv3(x)))
        
        # Final layer with Tanh (output in [-1, 1])
        x = torch.tanh(self.deconv4(x))
        
        return x


class VAE(nn.Module):
    """
    Variational Autoencoder (VAE)
    
    Combines Encoder and Decoder with reparameterization trick.
    
    Loss = Reconstruction Loss + β * KL Divergence
    """
    
    def __init__(self, latent_dim=128):
        super(VAE, self).__init__()
        
        self.latent_dim = latent_dim
        self.encoder = Encoder(latent_dim)
        self.decoder = Decoder(latent_dim)
        
    def reparameterize(self, mu, logvar):
        """
        Reparameterization trick: z = μ + σ * ε
        where ε ~ N(0, I)
        
        This allows gradients to flow through the sampling operation.
        """
        std = torch.exp(0.5 * logvar)  # σ = exp(0.5 * log(σ²))
        eps = torch.randn_like(std)     # ε ~ N(0, I)
        z = mu + eps * std              # z = μ + σ * ε
        return z
    
    def forward(self, x):
        # Encode
        mu, logvar = self.encoder(x)
        
        # Reparameterize
        z = self.reparameterize(mu, logvar)
        
        # Decode
        reconstruction = self.decoder(z)
        
        return reconstruction, mu, logvar
    
    def encode(self, x):
        """Encode image to latent mean (for inference)"""
        with torch.no_grad():
            mu, _ = self.encoder(x)
        return mu
    
    def decode(self, z):
        """Decode latent to image"""
        with torch.no_grad():
            return self.decoder(z)
    
    def generate(self, num_samples=16):
        """Generate new images from prior N(0, I)"""
        with torch.no_grad():
            z = torch.randn(num_samples, self.latent_dim).to(device)
            return self.decoder(z)

# %%
# Test model architecture
print("="*50)
print("MODEL ARCHITECTURE")
print("="*50)

test_model = VAE(latent_dim=LATENT_DIM).to(device)

# Count parameters
total_params = sum(p.numel() for p in test_model.parameters())
encoder_params = sum(p.numel() for p in test_model.encoder.parameters())
decoder_params = sum(p.numel() for p in test_model.decoder.parameters())

print(f"\nLatent dimension: {LATENT_DIM}")
print(f"Total parameters: {total_params:,}")
print(f"  Encoder: {encoder_params:,}")
print(f"  Decoder: {decoder_params:,}")

# Test forward pass
test_input = torch.randn(4, 3, 64, 64).to(device)
test_output, test_mu, test_logvar = test_model(test_input)

print(f"\nTest forward pass:")
print(f"  Input shape: {test_input.shape}")
print(f"  Output shape: {test_output.shape}")
print(f"  μ shape: {test_mu.shape}")
print(f"  logvar shape: {test_logvar.shape}")

del test_model  # Free memory

# %% [markdown]
# ---
# # LOSS FUNCTION AND TRAINING UTILITIES
# ---

# %%
def vae_loss(recon_x, x, mu, logvar, beta=1.0):
    """
    VAE Loss = Reconstruction Loss + β * KL Divergence
    
    Reconstruction Loss: MSE between input and reconstruction
    KL Divergence: KL(q(z|x) || p(z)) where p(z) = N(0, I)
    
    Args:
        recon_x: Reconstructed images
        x: Original images
        mu: Latent means
        logvar: Latent log-variances
        beta: Weight for KL term (β=1 for standard VAE, β>1 for β-VAE)
    
    Returns:
        total_loss, reconstruction_loss, kl_loss
    """
    # Reconstruction loss (MSE, summed over all pixels)
    recon_loss = F.mse_loss(recon_x, x, reduction='sum')
    
    # KL divergence: -0.5 * Σ(1 + log(σ²) - μ² - σ²)
    kl_loss = -0.5 * torch.sum(1 + logvar - mu.pow(2) - logvar.exp())
    
    # Total loss
    total_loss = recon_loss + beta * kl_loss
    
    return total_loss, recon_loss, kl_loss


def train_epoch(model, loader, optimizer, beta=1.0):
    """Train for one epoch"""
    model.train()
    total_loss = 0
    total_recon = 0
    total_kl = 0
    
    pbar = tqdm(loader, desc=f'Training (β={beta})')
    for data, _ in pbar:
        data = data.to(device)
        
        # Forward
        optimizer.zero_grad()
        recon, mu, logvar = model(data)
        
        # Loss
        loss, recon_loss, kl_loss = vae_loss(recon, data, mu, logvar, beta)
        
        # Backward
        loss.backward()
        optimizer.step()
        
        # Accumulate
        total_loss += loss.item()
        total_recon += recon_loss.item()
        total_kl += kl_loss.item()
        
        # Update progress bar
        pbar.set_postfix({
            'loss': f'{loss.item()/len(data):.1f}',
            'recon': f'{recon_loss.item()/len(data):.1f}',
            'kl': f'{kl_loss.item()/len(data):.1f}'
        })
    
    n = len(loader.dataset)
    return total_loss/n, total_recon/n, total_kl/n


def test_epoch(model, loader, beta=1.0):
    """Evaluate on test set"""
    model.eval()
    total_loss = 0
    total_recon = 0
    total_kl = 0
    
    with torch.no_grad():
        for data, _ in loader:
            data = data.to(device)
            recon, mu, logvar = model(data)
            loss, recon_loss, kl_loss = vae_loss(recon, data, mu, logvar, beta)
            
            total_loss += loss.item()
            total_recon += recon_loss.item()
            total_kl += kl_loss.item()
    
    n = len(loader.dataset)
    return total_loss/n, total_recon/n, total_kl/n


def tensor_to_image(tensor):
    """Convert tensor to numpy image for display"""
    img = denormalize(tensor).cpu().permute(1, 2, 0).numpy()
    return np.clip(img, 0, 1)

print("✅ Loss function and training utilities defined!")

# %% [markdown]
# ---
# # TASK 1: BASELINE VAE IMPLEMENTATION
# ## (Team Member: [Name])
# ---

# %% [markdown]
# ## Task 1: Train Baseline VAE (β=1.0)

# %%
print("="*70)
print("TASK 1: BASELINE VAE IMPLEMENTATION")
print("="*70)

# Initialize model
model = VAE(latent_dim=LATENT_DIM).to(device)
optimizer = optim.Adam(model.parameters(), lr=LEARNING_RATE)
scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='min', factor=0.5, patience=3)

# Training history
history = {
    'train_loss': [], 'train_recon': [], 'train_kl': [],
    'test_loss': [], 'test_recon': [], 'test_kl': []
}

best_test_loss = float('inf')

print(f"\nTraining baseline VAE (β=1.0)")
print(f"  Epochs: {TASK1_EPOCHS}")
print(f"  Learning rate: {LEARNING_RATE}")
print(f"  Batch size: {BATCH_SIZE}")
print("="*70)

# %%
# TRAINING LOOP
for epoch in range(1, TASK1_EPOCHS + 1):
    print(f"\n{'='*50}")
    print(f"Epoch {epoch}/{TASK1_EPOCHS}")
    print('='*50)
    
    # Train
    train_loss, train_recon, train_kl = train_epoch(model, train_loader, optimizer, beta=1.0)
    
    # Test
    test_loss, test_recon, test_kl = test_epoch(model, test_loader, beta=1.0)
    
    # Store history
    history['train_loss'].append(train_loss)
    history['train_recon'].append(train_recon)
    history['train_kl'].append(train_kl)
    history['test_loss'].append(test_loss)
    history['test_recon'].append(test_recon)
    history['test_kl'].append(test_kl)
    
    # Print summary
    print(f"\nTrain Loss: {train_loss:.2f} (Recon: {train_recon:.2f}, KL: {train_kl:.2f})")
    print(f"Test Loss:  {test_loss:.2f} (Recon: {test_recon:.2f}, KL: {test_kl:.2f})")
    
    # Learning rate scheduling
    scheduler.step(test_loss)
    
    # Save best model
    if test_loss < best_test_loss:
        best_test_loss = test_loss
        torch.save(model.state_dict(), '/kaggle/working/models/best_vae_model.pth')
        print(f"✅ Best model saved! (Loss: {best_test_loss:.2f})")

# Save final model
torch.save(model.state_dict(), '/kaggle/working/models/final_vae_model.pth')

print("\n" + "="*70)
print("TASK 1 TRAINING COMPLETE!")
print(f"Best test loss: {best_test_loss:.2f}")
print("="*70)

# %%
# Plot training curves
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

epochs_range = range(1, len(history['train_loss']) + 1)

# Total loss
axes[0].plot(epochs_range, history['train_loss'], 'b-', label='Train', linewidth=2)
axes[0].plot(epochs_range, history['test_loss'], 'r-', label='Test', linewidth=2)
axes[0].set_xlabel('Epoch')
axes[0].set_ylabel('Loss')
axes[0].set_title('Total Loss')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

# Reconstruction loss
axes[1].plot(epochs_range, history['train_recon'], 'b-', label='Train', linewidth=2)
axes[1].plot(epochs_range, history['test_recon'], 'r-', label='Test', linewidth=2)
axes[1].set_xlabel('Epoch')
axes[1].set_ylabel('Loss')
axes[1].set_title('Reconstruction Loss')
axes[1].legend()
axes[1].grid(True, alpha=0.3)

# KL divergence
axes[2].plot(epochs_range, history['train_kl'], 'b-', label='Train', linewidth=2)
axes[2].plot(epochs_range, history['test_kl'], 'r-', label='Test', linewidth=2)
axes[2].set_xlabel('Epoch')
axes[2].set_ylabel('Loss')
axes[2].set_title('KL Divergence')
axes[2].legend()
axes[2].grid(True, alpha=0.3)

plt.suptitle('Task 1: Baseline VAE Training Curves (β=1.0)', fontsize=14, y=1.02)
plt.tight_layout()
plt.savefig('/kaggle/working/results/task1/training_curves.png', dpi=150, bbox_inches='tight')
plt.show()

# %%
# Show reconstructions
model.eval()
test_data, _ = next(iter(test_loader))
test_data = test_data[:8].to(device)

with torch.no_grad():
    recon, _, _ = model(test_data)

fig, axes = plt.subplots(2, 8, figsize=(16, 4))
for i in range(8):
    axes[0, i].imshow(tensor_to_image(test_data[i]))
    axes[0, i].axis('off')
    if i == 0: axes[0, i].set_ylabel('Original', fontsize=12)
    
    axes[1, i].imshow(tensor_to_image(recon[i]))
    axes[1, i].axis('off')
    if i == 0: axes[1, i].set_ylabel('Reconstructed', fontsize=12)

plt.suptitle('Task 1: Baseline VAE Reconstructions', fontsize=14)
plt.tight_layout()
plt.savefig('/kaggle/working/results/task1/reconstructions.png', dpi=150, bbox_inches='tight')
plt.show()

# %%
# Generate new samples
samples = model.generate(64)
grid = make_grid(denormalize(samples), nrow=8)

plt.figure(figsize=(12, 12))
plt.imshow(grid.permute(1, 2, 0).cpu().numpy())
plt.axis('off')
plt.title('Task 1: Generated Samples from Prior N(0, I)', fontsize=14)
plt.savefig('/kaggle/working/results/task1/generated_samples.png', dpi=150, bbox_inches='tight')
plt.show()

# %%
# Save Task 1 summary
task1_summary = {
    'model': 'Baseline VAE',
    'latent_dim': LATENT_DIM,
    'image_size': IMG_SIZE,
    'beta': 1.0,
    'epochs': TASK1_EPOCHS,
    'learning_rate': LEARNING_RATE,
    'batch_size': BATCH_SIZE,
    'best_test_loss': float(best_test_loss),
    'final_test_loss': float(history['test_loss'][-1]),
    'total_parameters': sum(p.numel() for p in model.parameters()),
    'history': history
}

with open('/kaggle/working/results/task1/summary.json', 'w') as f:
    json.dump(task1_summary, f, indent=2)

print("\n" + "="*70)
print("✅ TASK 1 COMPLETE!")
print("="*70)
print("\nFiles saved:")
print("  📁 models/best_vae_model.pth")
print("  📁 models/final_vae_model.pth")
print("  📁 results/task1/training_curves.png")
print("  📁 results/task1/reconstructions.png")
print("  📁 results/task1/generated_samples.png")
print("  📁 results/task1/summary.json")
print("="*70)

# %% [markdown]
# ---
# # TASK 2: LATENT SPACE EXPLORATION
# ## (Team Member: [Name])
# ---

# %% [markdown]
# ## Task 2.1: Reconstruction Visualization (5+ images)

# %%
print("="*70)
print("TASK 2: LATENT SPACE EXPLORATION")
print("="*70)

# Load best model
baseline_model = VAE(latent_dim=LATENT_DIM).to(device)
baseline_model.load_state_dict(torch.load('/kaggle/working/models/best_vae_model.pth'))
baseline_model.eval()
print("✅ Loaded baseline model for Task 2")

# %%
# Task 2.1: Detailed Reconstruction Visualization
print("\n" + "-"*50)
print("Task 2.1: Reconstruction Visualization")
print("-"*50)

test_data, _ = next(iter(test_loader))
test_data = test_data[:8].to(device)

with torch.no_grad():
    recon, mu, logvar = baseline_model(test_data)

fig, axes = plt.subplots(3, 8, figsize=(16, 6))

for i in range(8):
    # Original
    axes[0, i].imshow(tensor_to_image(test_data[i]))
    axes[0, i].axis('off')
    if i == 0: axes[0, i].set_ylabel('Original', fontsize=11, rotation=0, labelpad=50)
    
    # Reconstructed
    axes[1, i].imshow(tensor_to_image(recon[i]))
    axes[1, i].axis('off')
    if i == 0: axes[1, i].set_ylabel('Reconstructed', fontsize=11, rotation=0, labelpad=50)
    
    # Difference (error map)
    diff = torch.abs(test_data[i] - recon[i]).mean(dim=0).cpu().numpy()
    axes[2, i].imshow(diff, cmap='hot')
    axes[2, i].axis('off')
    if i == 0: axes[2, i].set_ylabel('Difference', fontsize=11, rotation=0, labelpad=50)

plt.suptitle('Task 2.1: Reconstruction Visualization', fontsize=14, y=1.02)
plt.tight_layout()
plt.savefig('/kaggle/working/results/task2/reconstructions_detailed.png', dpi=150, bbox_inches='tight')
plt.show()

# Calculate metrics
mse = F.mse_loss(recon, test_data).item()
print(f"\nReconstruction MSE: {mse:.6f}")
print(f"Average |μ|: {mu.abs().mean().item():.4f}")
print(f"Average σ: {torch.exp(0.5 * logvar).mean().item():.4f}")

# %% [markdown]
# ## Task 2.2: Latent Interpolations (3+ pairs)

# %%
print("\n" + "-"*50)
print("Task 2.2: Latent Interpolations")
print("-"*50)

def interpolate(model, img1, img2, steps=10):
    """Interpolate between two images in latent space"""
    model.eval()
    with torch.no_grad():
        # Encode both images
        mu1, _ = model.encoder(img1.unsqueeze(0).to(device))
        mu2, _ = model.encoder(img2.unsqueeze(0).to(device))
        
        # Linear interpolation
        images = []
        for alpha in np.linspace(0, 1, steps):
            z = (1 - alpha) * mu1 + alpha * mu2
            img = model.decoder(z)
            images.append(img.squeeze(0))
    
    return images

# Get some test images
test_batch, _ = next(iter(test_loader))
test_batch = test_batch.to(device)

# Create interpolations for 4 pairs
num_pairs = 4
num_steps = 10

fig, axes = plt.subplots(num_pairs, num_steps, figsize=(num_steps * 1.5, num_pairs * 1.8))

for pair in range(num_pairs):
    img1 = test_batch[pair * 2]
    img2 = test_batch[pair * 2 + 1]
    
    interp_images = interpolate(baseline_model, img1, img2, num_steps)
    
    for step, img in enumerate(interp_images):
        axes[pair, step].imshow(tensor_to_image(img))
        axes[pair, step].axis('off')
        
        if pair == 0:
            alpha = step / (num_steps - 1)
            axes[pair, step].set_title(f'α={alpha:.1f}', fontsize=9)
    
    axes[pair, 0].set_ylabel(f'Pair {pair+1}', fontsize=10)

plt.suptitle('Task 2.2: Latent Space Interpolations\n(α=0: Image 1, α=1: Image 2)', fontsize=12, y=1.02)
plt.tight_layout()
plt.savefig('/kaggle/working/results/task2/interpolations.png', dpi=150, bbox_inches='tight')
plt.show()

print(f"✅ Generated {num_pairs} interpolation sequences with {num_steps} steps each")

# %% [markdown]
# ## Task 2.3: Latent Traversals

# %%
print("\n" + "-"*50)
print("Task 2.3: Latent Traversals")
print("-"*50)

def traverse_dimension(model, base_z, dim, range_min=-3, range_max=3, steps=9):
    """Traverse a single latent dimension"""
    model.eval()
    images = []
    values = np.linspace(range_min, range_max, steps)
    
    with torch.no_grad():
        for val in values:
            z = base_z.clone()
            z[0, dim] = val
            img = model.decoder(z)
            images.append(img.squeeze(0))
    
    return images, values

# Get a base image and encode it
base_img = test_batch[0:1]
with torch.no_grad():
    base_mu, _ = baseline_model.encoder(base_img)

# Traverse first 10 dimensions
num_dims = 10
num_steps = 9

fig, axes = plt.subplots(num_dims, num_steps, figsize=(num_steps * 1.2, num_dims * 1.3))

for dim in range(num_dims):
    images, values = traverse_dimension(baseline_model, base_mu, dim, -3, 3, num_steps)
    
    for step, (img, val) in enumerate(zip(images, values)):
        axes[dim, step].imshow(tensor_to_image(img))
        axes[dim, step].axis('off')
        
        if dim == 0:
            axes[dim, step].set_title(f'{val:.1f}', fontsize=8)
    
    axes[dim, 0].set_ylabel(f'Dim {dim}', fontsize=9, rotation=0, labelpad=25)

plt.suptitle('Task 2.3: Latent Dimension Traversals\n(Varying each dimension from -3 to +3)', fontsize=12, y=1.01)
plt.tight_layout()
plt.savefig('/kaggle/working/results/task2/traversals.png', dpi=150, bbox_inches='tight')
plt.show()

print(f"✅ Generated traversals for {num_dims} latent dimensions")

# %%
# Analyze latent dimension variance
print("\nAnalyzing latent dimension variance...")

all_mus = []
for batch_data, _ in test_loader:
    batch_data = batch_data.to(device)
    with torch.no_grad():
        mu, _ = baseline_model.encoder(batch_data)
        all_mus.append(mu.cpu())
    if len(all_mus) >= 10:  # Use 10 batches
        break

all_mus = torch.cat(all_mus, dim=0)
variance = all_mus.var(dim=0).numpy()

# Plot variance
fig, ax = plt.subplots(figsize=(14, 4))
ax.bar(range(len(variance)), variance, color='steelblue')
ax.set_xlabel('Latent Dimension', fontsize=11)
ax.set_ylabel('Variance', fontsize=11)
ax.set_title('Task 2.3: Variance per Latent Dimension', fontsize=12)
ax.grid(True, alpha=0.3, axis='y')

plt.tight_layout()
plt.savefig('/kaggle/working/results/task2/latent_variance.png', dpi=150, bbox_inches='tight')
plt.show()

# Print top dimensions
top_dims = np.argsort(variance)[::-1][:10]
print("\nTop 10 most variable dimensions:")
for i, dim in enumerate(top_dims):
    print(f"  {i+1}. Dimension {dim}: variance = {variance[dim]:.4f}")

# %%
print("\n" + "="*70)
print("✅ TASK 2 COMPLETE!")
print("="*70)
print("\nFiles saved:")
print("  📁 results/task2/reconstructions_detailed.png")
print("  📁 results/task2/interpolations.png")
print("  📁 results/task2/traversals.png")
print("  📁 results/task2/latent_variance.png")
print("="*70)

# %% [markdown]
# ---
# # TASK 3: β-VAE VARIANT
# ## (Team Member: [Name])
# ---

# %%
print("="*70)
print("TASK 3: β-VAE VARIANT")
print("="*70)
print(f"\nWill train VAEs with β values: {BETA_VALUES}")
print(f"Epochs per model: {TASK3_EPOCHS}")

# %%
def train_beta_vae(beta, epochs):
    """Train a VAE with specified β value"""
    print(f"\n{'='*50}")
    print(f"Training β-VAE with β = {beta}")
    print(f"{'='*50}")
    
    # Initialize new model
    model = VAE(latent_dim=LATENT_DIM).to(device)
    optimizer = optim.Adam(model.parameters(), lr=LEARNING_RATE)
    scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, 'min', factor=0.5, patience=3)
    
    history = {
        'train_loss': [], 'train_recon': [], 'train_kl': [],
        'test_loss': [], 'test_recon': [], 'test_kl': []
    }
    best_loss = float('inf')
    
    for epoch in range(1, epochs + 1):
        # Train
        train_loss, train_recon, train_kl = train_epoch(model, train_loader, optimizer, beta)
        
        # Test
        test_loss, test_recon, test_kl = test_epoch(model, test_loader, beta)
        
        # Store
        history['train_loss'].append(train_loss)
        history['train_recon'].append(train_recon)
        history['train_kl'].append(train_kl)
        history['test_loss'].append(test_loss)
        history['test_recon'].append(test_recon)
        history['test_kl'].append(test_kl)
        
        print(f"Epoch {epoch}/{epochs}: Test Loss={test_loss:.2f} (Recon:{test_recon:.2f}, KL:{test_kl:.2f})")
        
        scheduler.step(test_loss)
        
        if test_loss < best_loss:
            best_loss = test_loss
            torch.save(model.state_dict(), f'/kaggle/working/models/vae_beta{beta}.pth')
    
    # Load best
    model.load_state_dict(torch.load(f'/kaggle/working/models/vae_beta{beta}.pth'))
    
    return model, history

# %%
# Train β-VAE variants
beta_models = {1.0: baseline_model}  # Include baseline
beta_histories = {1.0: history}       # Baseline history from Task 1

for beta in BETA_VALUES:
    model, hist = train_beta_vae(beta, TASK3_EPOCHS)
    beta_models[beta] = model
    beta_histories[beta] = hist

print("\n✅ All β-VAE models trained!")

# %%
# Compare reconstructions
print("\n" + "-"*50)
print("Task 3: Comparing Reconstruction Quality")
print("-"*50)

test_imgs, _ = next(iter(test_loader))
test_imgs = test_imgs[:6].to(device)

betas = sorted(beta_models.keys())
num_imgs = 6

fig, axes = plt.subplots(len(betas) + 1, num_imgs, figsize=(num_imgs * 2, (len(betas) + 1) * 2))

# Original images (first row)
for i in range(num_imgs):
    axes[0, i].imshow(tensor_to_image(test_imgs[i]))
    axes[0, i].axis('off')
axes[0, 0].set_ylabel('Original', fontsize=10, rotation=0, labelpad=50)

# Reconstructions for each β
mse_scores = {}
for row, beta in enumerate(betas):
    model = beta_models[beta]
    model.eval()
    
    with torch.no_grad():
        recon, _, _ = model(test_imgs)
    
    mse = F.mse_loss(recon, test_imgs).item()
    mse_scores[beta] = mse
    
    for i in range(num_imgs):
        axes[row + 1, i].imshow(tensor_to_image(recon[i]))
        axes[row + 1, i].axis('off')
    
    axes[row + 1, 0].set_ylabel(f'β={beta}\nMSE={mse:.2f}', fontsize=10, rotation=0, labelpad=60)

plt.suptitle('Task 3: Reconstruction Quality vs β Value', fontsize=14, y=1.02)
plt.tight_layout()
plt.savefig('/kaggle/working/results/task3/reconstruction_comparison.png', dpi=150, bbox_inches='tight')
plt.show()

print("\nReconstruction MSE by β:")
for beta, mse in sorted(mse_scores.items()):
    print(f"  β={beta}: MSE = {mse:.4f}")

# %%
# Compare traversals
print("\n" + "-"*50)
print("Task 3: Comparing Latent Disentanglement")
print("-"*50)

base_img = test_imgs[0:1]
dims_to_show = [0, 1, 2, 3, 4]
num_steps = 7

fig, axes = plt.subplots(len(betas) * len(dims_to_show), num_steps,
                         figsize=(num_steps * 1.2, len(betas) * len(dims_to_show) * 1.1))

row = 0
for beta in betas:
    model = beta_models[beta]
    model.eval()
    
    with torch.no_grad():
        base_mu, _ = model.encoder(base_img)
    
    for dim in dims_to_show:
        images, values = traverse_dimension(model, base_mu, dim, -3, 3, num_steps)
        
        for col, (img, val) in enumerate(zip(images, values)):
            axes[row, col].imshow(tensor_to_image(img))
            axes[row, col].axis('off')
            
            if row == 0:
                axes[row, col].set_title(f'{val:.1f}', fontsize=8)
        
        axes[row, 0].set_ylabel(f'β={beta}\nd{dim}', fontsize=8, rotation=0, labelpad=35)
        row += 1

plt.suptitle('Task 3: Latent Traversal Comparison\n(Better disentanglement = more distinct changes per dimension)', 
             fontsize=11, y=1.01)
plt.tight_layout()
plt.savefig('/kaggle/working/results/task3/traversal_comparison.png', dpi=150, bbox_inches='tight')
plt.show()

# %%
# KL per dimension analysis
print("\n" + "-"*50)
print("Task 3: KL Divergence per Dimension")
print("-"*50)

fig, axes = plt.subplots(1, len(betas), figsize=(5 * len(betas), 4))

for ax, beta in zip(axes, betas):
    model = beta_models[beta]
    model.eval()
    
    all_kl = []
    for data, _ in test_loader:
        data = data.to(device)
        with torch.no_grad():
            mu, logvar = model.encoder(data)
            kl_per_dim = -0.5 * (1 + logvar - mu.pow(2) - logvar.exp())
            all_kl.append(kl_per_dim.mean(dim=0).cpu())
        if len(all_kl) >= 10:
            break
    
    avg_kl = torch.stack(all_kl).mean(dim=0).numpy()
    active_dims = (avg_kl > 0.1).sum()
    
    ax.bar(range(len(avg_kl)), avg_kl, color='steelblue')
    ax.set_xlabel('Dimension')
    ax.set_ylabel('KL Divergence')
    ax.set_title(f'β = {beta}\nActive dims: {active_dims}')
    ax.axhline(y=0.1, color='red', linestyle='--', alpha=0.5, label='Threshold')
    ax.grid(True, alpha=0.3, axis='y')

plt.suptitle('Task 3: KL per Dimension (Higher β → More Concentrated → Better Disentanglement)', 
             fontsize=11, y=1.05)
plt.tight_layout()
plt.savefig('/kaggle/working/results/task3/kl_analysis.png', dpi=150, bbox_inches='tight')
plt.show()

# %%
# Final summary
print("\n" + "="*70)
print("TASK 3: FINAL COMPARISON SUMMARY")
print("="*70)

print(f"\n{'β Value':<10} {'Recon MSE':<15} {'Test Loss':<15}")
print("-" * 40)

for beta in betas:
    mse = mse_scores[beta]
    final_loss = beta_histories[beta]['test_loss'][-1] if beta_histories[beta] else 'N/A'
    print(f"{beta:<10} {mse:<15.4f} {final_loss:<15.2f}")

# %%
# Save Task 3 summary
task3_summary = {
    'beta_values': list(betas),
    'epochs_per_model': TASK3_EPOCHS,
    'mse_scores': {str(k): v for k, v in mse_scores.items()},
    'histories': {str(k): v for k, v in beta_histories.items()}
}

with open('/kaggle/working/results/task3/summary.json', 'w') as f:
    json.dump(task3_summary, f, indent=2)

print("\n" + "="*70)
print("✅ TASK 3 COMPLETE!")
print("="*70)
print("\nFiles saved:")
print("  📁 models/vae_beta4.pth")
print("  📁 models/vae_beta10.pth")
print("  📁 results/task3/reconstruction_comparison.png")
print("  📁 results/task3/traversal_comparison.png")
print("  📁 results/task3/kl_analysis.png")
print("  📁 results/task3/summary.json")
print("="*70)

# %% [markdown]
# ---
# # FINAL SUMMARY
# ---

# %%
print("\n" + "="*70)
print("🎉 ALL TASKS COMPLETE! 🎉")
print("="*70)

print("""
TASK 1: Baseline VAE Implementation ✅
  - Built and trained VAE on CelebA
  - Architecture: Conv encoder/decoder, latent_dim=128
  - Training curves documented
  - Best model saved

TASK 2: Latent Space Exploration ✅
  - Reconstructions visualized (8 images)
  - Latent interpolations (4 pairs)
  - Latent traversals (10 dimensions)
  - Variance analysis completed

TASK 3: β-VAE Variant ✅
  - Trained β=4 and β=10 variants
  - Reconstruction quality compared
  - Disentanglement analyzed
  - Trade-offs discussed

OUTPUT FILES:
  📁 /kaggle/working/models/
     - best_vae_model.pth (baseline)
     - final_vae_model.pth
     - vae_beta4.pth
     - vae_beta10.pth
  
  📁 /kaggle/working/results/task1/
     - training_curves.png
     - reconstructions.png
     - generated_samples.png
     - summary.json
  
  📁 /kaggle/working/results/task2/
     - reconstructions_detailed.png
     - interpolations.png
     - traversals.png
     - latent_variance.png
  
  📁 /kaggle/working/results/task3/
     - reconstruction_comparison.png
     - traversal_comparison.png
     - kl_analysis.png
     - summary.json
""")

print("="*70)
print("Don't forget to download all files from /kaggle/working/")
print("Click 'Save Version' → 'Quick Save' to save your notebook")
print("="*70)

# %% [markdown]
# ---
# # Discussion: β-VAE Trade-offs
# 
# ## Key Observations:
# 
# ### 1. Reconstruction Quality
# - **β=1 (Standard VAE)**: Best reconstruction quality, lowest MSE
# - **β=4**: Slightly blurrier reconstructions
# - **β=10**: Noticeably blurrier, some details lost
# 
# ### 2. Latent Disentanglement
# - **β=1**: Entangled - dimensions often affect multiple features
# - **β=4**: Improved disentanglement - some dimensions control single features
# - **β=10**: Better disentanglement - fewer active dimensions, more independent
# 
# ### 3. The Fundamental Trade-off
# - Higher β → More pressure on KL term → Forces simpler latent distribution
# - Simpler distribution → Fewer active dimensions → Better disentanglement
# - BUT: Information bottleneck → Worse reconstruction
# 
# ### 4. Practical Recommendations
# - **For generation tasks**: Use β=1 (best quality)
# - **For interpretable representations**: Use β=4-10
# - **For faces (CelebA)**: β=4 often provides good balance
# 
# ---
```
